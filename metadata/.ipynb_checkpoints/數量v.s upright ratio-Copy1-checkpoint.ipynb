```python
import matplotlib.pyplot as plt

# === 資料設定 ===
x = [10, 50, 100, 200]  # 分子數量
upright_sds = [0.1368, 0.3506, 0.5385, 0.6391]
upright_nia = [0.4606, 0.5162, 0.6196, 0.6610]

# === 繪圖 ===
plt.figure(figsize=(8,6))

plt.plot(x, upright_sds, marker='o', color='blue', linewidth=2, label='SDS')
plt.plot(x, upright_nia, marker='s', color='red', linewidth=2, label='Niaproof')

# === 圖形設定 ===
plt.xlabel('Number of Molecules', fontsize=24)
plt.ylabel('Upright Ratio', fontsize=24)
plt.tick_params(axis='both', which='major', labelsize=18)

#plt.title('Upright Ratio vs Number of Molecules', fontsize=24)
plt.legend(fontsize=24, loc='lower right')


plt.tight_layout()
plt.show()
```
